# 🚀 DEMO LAUNCHER — Stock Forecasting Top 5 VN30

Notebook này khởi động **dashboard Streamlit** ở cả 2 môi trường:

| Môi trường | Cách hoạt động |
|---|---|
| **Google Colab** | Tự gắn Drive → cài thư viện → chạy Streamlit background → tạo **URL public qua localtunnel** (mật khẩu truy cập = IP máy Colab, in ngay dưới cell) |
| **Local (VS Code)** | In đúng lệnh cần chạy trong terminal (`.venv` sẵn sàng, không cài lại) |

> **Yêu cầu Colab:** project phải nằm trong Drive, mặc định tại `/content/drive/MyDrive/Stock_Forecasting_Project` — nếu khác hãy sửa biến `COLAB_PROJECT_PATH` ở cell dưới.

```python
# ==== BUOC 1: DO MOI TRUONG + XAC DINH DUONG DAN PROJECT ====
import sys
from pathlib import Path

COLAB_PROJECT_PATH = '/content/drive/MyDrive/Stock_Forecasting_Project'

IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive')
    PROJECT_PATH = Path(COLAB_PROJECT_PATH)
else:
    p = Path.cwd()
    while not (p / '01_Data').exists() and p != p.parent:
        p = p.parent
    PROJECT_PATH = p

APP = PROJECT_PATH / '04_Source_Code' / 'app.py'
assert APP.exists(), f'Khong thay app.py tai {APP} — kiem tra COLAB_PROJECT_PATH!'
print('IN_COLAB      :', IN_COLAB)
print('PROJECT_PATH  :', PROJECT_PATH)
print('app.py        :', APP)
```

```python
# ==== BUOC 2: CAI THU VIEN (chi Colab — local da co .venv) ====
if IN_COLAB:
    get_ipython().system('pip -q install pandas numpy scikit-learn matplotlib torch streamlit plotly joblib yfinance')
    print('Da cai xong thu vien cho Colab.')
else:
    print('Local: bo qua (dung .venv san co).')
```

```python
# ==== BUOC 3: KHOI DONG DEMO ====
import time

if IN_COLAB:
    get_ipython().system_raw(f"streamlit run '{APP}' --server.port 8599 --server.headless true > /content/st.log 2>&1 &")
    time.sleep(6)
    get_ipython().system_raw('npx localtunnel --port 8599 > /content/lt.log 2>&1 &')
    time.sleep(8)
    url = open('/content/lt.log').read().strip()
    import urllib.request
    ip = urllib.request.urlopen('https://api.ipify.org').read().decode()
    print('=' * 55)
    print('PUBLIC URL :', url if url else '(chua co — chay lai cell nay sau 15 giay)')
    print('TUNNEL PASSWORD (IP):', ip)
    print('Mo URL -> dan IP vao o Tunnel password -> Enjoy!')
    print('=' * 55)
    print('Log Streamlit: /content/st.log | Log tunnel: /content/lt.log')
else:
    print('Ban dang o LOCAL (VS Code). Mo terminal tai thu muc goc project va chay:')
    print('    streamlit run 04_Source_Code/app.py --server.port 8599')
    print('Hoac muon chay ngay trong cell nay (se block):')
    print('    !streamlit run 04_Source_Code/app.py --server.port 8599')
    print('Dashboard: http://localhost:8599')
```
